```python
def load_job_data(file_path):
    return pd.read_csv(
        file_path,
        sep="\t",
        encoding="utf-16"
    )

march_df = load_job_data(
    "../data/raw/March 2026 Data Set/job-bank-open-data-all-job-postings-en-mar2026.csv"
)
march_df.shape

may_df = load_job_data(
    "../data/raw/May 2026 Data Set/job-bank-open-data-all-job-postings-en-may2026 (1).csv"
)
may_df.shape

june_df = load_job_data(
    "../data/raw/June 2026 Data Set/job-bank-open-data-all-job-postings-en-june2026 (1).csv"
)
june_df.shape

july_df = load_job_data(
    "../data/raw/July 2026 Data Set/job-bank-open-data-all-job-postings-en-july2026 (1).csv"
)
july_df.shape
```

```python
def filter_data_jobs(df):
    keywords = (
        "data analyst|"
        "data scientist|"
        "business intelligence|"
        "bi analyst|"
        "analytics analyst|"
        "data analytics|"
        "data quality analyst|"
        "reporting analyst"
    )

    filtered = df[
        df["Job Title"].str.contains(
            keywords,
            case=False,
            na=False
        )
    ].copy()

    filtered = filtered[
        filtered["Job Title"].str.lower() != "data analytics manager"
    ].copy()

    return filtered

march_jobs = filter_data_jobs(march_df)
march_jobs.shape

april_jobs = filter_data_jobs(df)
april_jobs.shape

may_jobs = filter_data_jobs(may_df)
may_jobs.shape

june_jobs = filter_data_jobs(june_df)
june_jobs.shape

july_jobs = filter_data_jobs(july_df)
july_jobs.shape
```

```python
def annual_salary(value, salary_per):
    if salary_per == "Year":
        return value
    
    elif salary_per == "Hour":
        return value * 40 * 52
    
    elif salary_per == "Bi-weekly":
        return value * 26
    
    else:
        return None

final_jobs["Estimated_Annual_Min"] = final_jobs.apply(
    lambda row: annual_salary(
        row["Salary_Minimum"],
        row["Salary_Per"]
    ),
    axis=1
)

final_jobs["Estimated_Annual_Max"] = final_jobs.apply(
    lambda row: annual_salary(
        row["Salary_Maximum"],
        row["Salary_Per"]
    ),
    axis=1
)

final_jobs["Estimated_Annual_Mid"] = (
    final_jobs["Estimated_Annual_Min"] +
    final_jobs["Estimated_Annual_Max"]
) / 2

final_jobs[
    [
        "Job_Title",
        "Salary_Per",
        "Salary_Minimum",
        "Salary_Maximum",
        "Estimated_Annual_Min",
        "Estimated_Annual_Max",
        "Estimated_Annual_Mid"
    ]
].head(15)
```

```python
import pandas as pd
import csv

file_path = "../data/raw/OaSIS Skills/skills_oasis_2025_v1.1.csv"

with open(file_path, "r", encoding="utf-8-sig") as file:
    sample = file.read(5000)


detected_separator = csv.Sniffer().sniff(
    sample,
    delimiters=",;\t|"
).delimiter

print("Detected separator:", repr(detected_separator))


oasis_skills = pd.read_csv(
    file_path,
    sep=detected_separator,
    encoding="utf-8-sig"
)

print("\nDataset shape:")
print(oasis_skills.shape)

print("\nColumn names:")
print(oasis_skills.columns.tolist())

print("\nFirst 5 rows:")
display(oasis_skills.head())


print("Job Bank NOC21 codes:")
print(
    final_jobs["NOC21_Code"]
    .dropna()
    .astype(int)
    .astype(str)
    .unique()[:20]
)

print("\nOaSIS codes:")
print(
    oasis_skills["Code OaSIS"]
    .dropna()
    .astype(str)
    .head(20)
    .tolist()
)


oasis_skills["OaSIS_Code_String"] = (
    oasis_skills["Code OaSIS"]
    .astype(str)
)


oasis_skills["NOC21_Code_Clean"] = (
    oasis_skills["OaSIS_Code_String"]
    .str.split(".")
    .str[0]
    .str.zfill(5)
)


final_jobs["NOC21_Code_Clean"] = (
    final_jobs["NOC21_Code"]
    .astype("Int64")
    .astype(str)
    .str.zfill(5)
)



print("Job Bank NOC codes:")

print(
    final_jobs["NOC21_Code_Clean"]
    .dropna()
    .unique()
)


matching_oasis = oasis_skills[
    oasis_skills["NOC21_Code_Clean"].isin(
        final_jobs["NOC21_Code_Clean"].dropna().unique()
    )
].copy()


print("\nNumber of matching OaSIS profiles:")
print(len(matching_oasis))


print("\nMatching occupations:")

display(
    matching_oasis[
        [
            "Code OaSIS",
            "NOC21_Code_Clean",
            "OaSIS 2021 Labels"
        ]
    ]
)


selected_skills = [
    "Digital Literacy",
    "Numeracy",
    "Problem Solving",
    "Systems Analysis",
    "Critical Thinking",
    "Decision Making",
    "Reading Comprehension",
    "Writing",
    "Troubleshooting",
    "Time Management"
]



relevant_oasis = oasis_skills[
    oasis_skills["NOC21_Code_Clean"].isin(
        final_jobs["NOC21_Code_Clean"].dropna().unique()
    )
].copy()



noc_skill_profiles = (
    relevant_oasis
    .groupby("NOC21_Code_Clean")[selected_skills]
    .mean()
    .reset_index()
)


print("Number of NOC skill profiles:")
print(noc_skill_profiles.shape)


print("\nSkill profiles:")

display(noc_skill_profiles)



jobs_with_skills = final_jobs.merge(
    noc_skill_profiles,
    on="NOC21_Code_Clean",
    how="left"
)


print("\nOriginal job dataset:")
print(final_jobs.shape)

print("\nDataset after adding skills:")
print(jobs_with_skills.shape)



print("\nNumber of rows before merge:")
print(len(final_jobs))

print("\nNumber of rows after merge:")
print(len(jobs_with_skills))



print("\nJobs with OaSIS skill information:")

print(
    jobs_with_skills["Digital Literacy"]
    .notna()
    .sum()
)


# --------------------------------------------------
# 6. Calculate average skill ratings across
#    the occupations represented by our postings
# --------------------------------------------------

average_skills = (
    jobs_with_skills[selected_skills]
    .mean()
    .sort_values(ascending=False)
)


print("\nAverage occupational skill ratings:")

display(average_skills)
```

```python


def add_job_match(
    wic_id,
    job_bank_number,
    job_bank_url,
    technologies,
    status="Verified"
):
    """
    Adds verified Job Bank information to one posting
    in technology_collection.
    """

    mask = technology_collection["WIC_ID"] == wic_id

    technology_collection.loc[
        mask,
        "Job_Bank_Job_Number"
    ] = str(job_bank_number)

    technology_collection.loc[
        mask,
        "Job_Bank_URL"
    ] = job_bank_url

    technology_collection.loc[
        mask,
        "Technology_Knowledge"
    ] = technologies

    technology_collection.loc[
        mask,
        "Match_Status"
    ] = status


def save_technology_progress():

    technology_collection.to_csv(
        "../data/processed/technology_collection.csv",
        index=False
    )

    print("Progress saved!")

    print("\nMATCH STATUS")
    print(
        technology_collection[
            "Match_Status"
        ].value_counts()
    )
```

```python


import pandas as pd
import re


verified_technology_jobs = technology_collection[
    technology_collection["Match_Status"].isin(
        [
            "Verified",
            "Verified - No technology listed"
        ]
    )
].copy()

print("Verified direct Job Bank postings:")
print(len(verified_technology_jobs))



verified_technology_jobs["Technology_Knowledge"] = (
    verified_technology_jobs["Technology_Knowledge"]
    .fillna("")
    .astype(str)
)


technology_patterns = {

    "Python": [
        r"\bpython\b"
    ],

    # Only count SQL when SQL itself is explicitly mentioned
    "SQL": [
        r"(?<![a-zA-Z])sql(?![a-zA-Z])"
    ],

    "MySQL": [
        r"\bmysql\b"
    ],

    "Excel": [
        r"\bexcel\b",
        r"\bmicrosoft excel\b",
        r"\bms excel\b"
    ],

    "Power BI": [
        r"\bpower\s*bi\b",
        r"\bpowerbi\b"
    ],

    "Tableau": [
        r"\btableau\b"
    ],

    "R": [
        r"\br programming\b",
        r"\br language\b"
    ],

    "AWS": [
        r"\baws\b",
        r"\bamazon web services\b"
    ],

    "Azure": [
        r"\bazure\b",
        r"\bmicrosoft azure\b"
    ],

    "C": [
        r"(?<!\+)\bc\b(?!\+)"
    ],

    # Correct pattern for C++
    "C++": [
        r"(?<!\w)c\+\+(?!\w)"
    ],

    "Java": [
        r"\bjava\b"
    ],

    "Spark": [
        r"\bapache spark\b",
        r"\bspark\b"
    ],

    "Databricks": [
        r"\bdatabricks\b"
    ],

    "Snowflake": [
        r"\bsnowflake\b"
    ],

    "SAS": [
        r"\bsas\b"
    ],

    "SASS": [
        r"\bsass\b"
    ],

    "Linux": [
        r"\blinux\b"
    ],

    "Git": [
        r"\bgit\b",
        r"\bgithub\b",
        r"\bgitlab\b"
    ],

    "Machine Learning": [
        r"\bmachine learning\b"
    ],

    "Big Data": [
        r"\bbig data\b"
    ],

    "Data Warehouse": [
        r"\bdata warehouse\b"
    ],

    "DB2": [
        r"\bdb2\b"
    ],

    "Visual Studio": [
        r"\bvisual studio\b"
    ],

    "ERP Software": [
        r"\benterprise resource planning\b",
        r"\berp\b"
    ]
}



for technology, patterns in technology_patterns.items():

    combined_pattern = "|".join(patterns)

    verified_technology_jobs[technology] = (
        verified_technology_jobs["Technology_Knowledge"]
        .str.contains(
            combined_pattern,
            case=False,
            regex=True,
            na=False
        )
        .astype(int)
    )



technology_results = []

total_verified = len(verified_technology_jobs)


for technology in technology_patterns.keys():

    mentions = int(
        verified_technology_jobs[technology].sum()
    )

    if total_verified > 0:
        percentage = (
            mentions / total_verified
        ) * 100
    else:
        percentage = 0

    technology_results.append(
        {
            "Technology": technology,
            "Postings_Mentioning": mentions,
            "Demand_Percentage": round(
                percentage,
                2
            )
        }
    )


technology_demand = pd.DataFrame(
    technology_results
)


technology_demand = (
    technology_demand
    .sort_values(
        [
            "Postings_Mentioning",
            "Technology"
        ],
        ascending=[
            False,
            True
        ]
    )
    .reset_index(drop=True)
)


technology_demand_nonzero = (
    technology_demand[
        technology_demand[
            "Postings_Mentioning"
        ] > 0
    ]
    .reset_index(drop=True)
)



print("\nTECHNOLOGY DEMAND")

print(
    "Denominator:",
    total_verified,
    "verified direct Job Bank postings"
)

display(
    technology_demand_nonzero
)



verified_technology_jobs.to_csv(
    "../data/processed/technology_mentions.csv",
    index=False
)

technology_demand.to_csv(
    "../data/processed/technology_demand.csv",
    index=False
)


print("\nSaved:")
print("technology_mentions.csv")
print("technology_demand.csv")
```

```python


import pandas as pd



verification_queue = technology_collection[
    ~technology_collection["Match_Status"].isin(
        [
            "Verified",
            "Verified - No technology listed"
        ]
    )
].copy()


priority_order = {
    "Needs technology review": 1,
    "Needs verification": 2,
    "Not checked": 3
}

verification_queue["Priority"] = (
    verification_queue["Match_Status"]
    .map(priority_order)
)


verification_queue = (
    verification_queue
    .sort_values(
        ["Priority", "Posting_Date"]
    )
    .reset_index(drop=True)
)



def build_search_query(row):

    parts = [
        str(row["Job_Title"])
    ]

    if pd.notna(row["City"]):
        parts.append(str(row["City"]))

    if pd.notna(row["Province"]):
        parts.append(str(row["Province"]))

    parts.append("Job Bank Canada")

    if pd.notna(row["Salary_Minimum"]):
        parts.append(str(row["Salary_Minimum"]))

    return " ".join(parts)


verification_queue["Search_Query"] = (
    verification_queue.apply(
        build_search_query,
        axis=1
    )
)



print("=" * 70)
print("FINAL TECHNOLOGY VERIFICATION QUEUE")
print("=" * 70)

print(
    "\nJobs still requiring work:",
    len(verification_queue)
)

print("\nSTATUS:")
print(
    verification_queue[
        "Match_Status"
    ].value_counts()
)


print("\nNEXT 10 JOBS TO RESOLVE:")

display(
    verification_queue[
        [
            "WIC_ID",
            "Job_Title",
            "Posting_Date",
            "Province",
            "City",
            "Salary_Per",
            "Salary_Minimum",
            "Salary_Maximum",
            "Job_Bank_Job_Number",
            "Match_Status",
            "Search_Query"
        ]
    ].head(10)
)
```

```python

import pandas as pd
from collections import Counter



verified_technology_jobs = technology_collection[
    technology_collection["Match_Status"].isin(
        [
            "Verified",
            "Verified - No technology listed"
        ]
    )
].copy()


print(
    "Verified postings used for technology analysis:",
    len(verified_technology_jobs)
)



technology_name_map = {

    # Cloud
    "Amazon Web Services (AWS)": "AWS",
    "Amazon Web Services": "AWS",

    # Microsoft
    "Microsoft Office": "MS Office",
    "MS Office": "MS Office",
    "Microsoft Excel": "Excel",
    "MS Excel": "Excel",
    "Microsoft Power BI": "Power BI",

    # Programming
    "Programming languages": "Programming Languages",
    "Programming software": "Programming Software",

    # Machine learning
    "Machine learning": "Machine Learning",

    # Data
    "Data analysis software": "Data Analysis Software",
    "Data Warehouse": "Data Warehouse",

    # Development
    "Software development": "Software Development",

    # Operating systems
    "MS Windows": "Windows",

    # Database
    "Database software": "Database Software",

    # Network
    "Networking software": "Networking Software",
    "Networking security": "Network Security",

    # Other
    "Spreadsheet": "Spreadsheet Software"
}



technology_rows = []


for _, row in verified_technology_jobs.iterrows():

    technology_text = row["Technology_Knowledge"]

    # Skip postings where Job Bank explicitly listed no technology
    if pd.isna(technology_text):
        continue

    technology_text = str(technology_text).strip()

    if technology_text == "":
        continue


    technologies = technology_text.split(";")


    # Avoid counting the same technology twice
    # within one job posting
    technologies_for_job = set()


    for technology in technologies:

        technology = technology.strip()

        if not technology:
            continue


        # Standardize names
        technology = technology_name_map.get(
            technology,
            technology
        )


        technologies_for_job.add(
            technology
        )


    for technology in technologies_for_job:

        technology_rows.append(
            {
                "WIC_ID": row["WIC_ID"],
                "Job_Title": row["Job_Title"],
                "Posting_Date": row["Posting_Date"],
                "Province": row["Province"],
                "Technology": technology
            }
        )



technology_mentions = pd.DataFrame(
    technology_rows
)


print(
    "\nTechnology mentions:",
    len(technology_mentions)
)


print(
    "Unique technologies:",
    technology_mentions["Technology"].nunique()
)



verified_denominator = len(
    verified_technology_jobs
)


technology_demand = (
    technology_mentions
    .groupby("Technology")["WIC_ID"]
    .nunique()
    .reset_index(
        name="Postings_Mentioning"
    )
)


technology_demand[
    "Demand_Percentage"
] = (

    technology_demand[
        "Postings_Mentioning"
    ]

    / verified_denominator

    * 100
).round(2)


technology_demand = (
    technology_demand
    .sort_values(
        [
            "Postings_Mentioning",
            "Technology"
        ],
        ascending=[
            False,
            True
        ]
    )
    .reset_index(drop=True)
)



print("\n" + "=" * 60)

print("FINAL TECHNOLOGY DEMAND")

print("=" * 60)

print(
    "\nDenominator:",
    verified_denominator,
    "verified direct Job Bank postings"
)


display(
    technology_demand.head(20)
)



technology_mentions.to_csv(
    "../data/processed/technology_mentions.csv",
    index=False
)


technology_demand.to_csv(
    "../data/processed/technology_demand.csv",
    index=False
)


verified_technology_jobs.to_csv(
    "../data/processed/verified_technology_jobs.csv",
    index=False
)


print("\nFILES SAVED:")

print(
    "technology_mentions.csv"
)

print(
    "technology_demand.csv"
)

print(
    "verified_technology_jobs.csv"
)
```
